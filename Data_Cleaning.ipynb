#Data Cleaning

```python
#to remove timestamp and gmail from Cleaned Data sheet
clean_df = clean_df.drop(columns=['Timestamp', 'Email'])

print("Rows after removing columns:", len(clean_df))
```

```python
#to clean spaces in text
for col in clean_df.columns:
    if clean_df[col].dtype == 'object':
        clean_df[col] = clean_df[col].apply(
            lambda x: x.strip() if isinstance(x, str) else x
        )
```

```python
#to remark Mandatory questions
mandatory_cols = [
    'Q1. Age',
    'Q2. Province',
    'Q3. District',
    'Q4. Local Authority',
    'Q5. Area Type',
    'Q6. Place Type',
    'Q7. No. of People',
    'Q8. Land Size',
    'Q9. Waste Storage Method',
    'Q14. Separate Recyclables?',
    'Q15. Recycle Waste?',
    'Q19. Distance to Collection Point',
    'Q20. Collection Frequency',
    'Q21. Disposal Method',
    'Q22. Monthly Spend (Rs.)',
    'Q23. Biggest Problem',
    'Q24. Use Mobile App?',
    'Q25. Satisfaction'
]
```

```python
for col in mandatory_cols:
    clean_df[col] = clean_df[col].replace(r'^\s*$', np.nan, regex=True)

clean_df = clean_df.dropna(subset=mandatory_cols)

print("Rows after mandatory-field cleaning:", len(clean_df))
```

```python
error_rows = clean_df[
    clean_df[mandatory_cols].isna().any(axis=1)
]

print("Rows containing errors:")
print(error_rows[mandatory_cols])

print("\nOriginal row numbers:")
print(error_rows.index.tolist())
```

```python
#q16 contain answers like "No", "-". to normalize it
clean_df['Q16. How Recycle?'] = clean_df['Q16. How Recycle?'].replace(
    ['-', 'No'],
    np.nan
)
```

```python
#to Q15 → Q16 validation
q16_invalid = (
    ((clean_df['Q15. Recycle Waste?'] == 'Yes') &
     (clean_df['Q16. How Recycle?'].isna())) |

    ((clean_df['Q15. Recycle Waste?'] == 'No') &
     (clean_df['Q16. How Recycle?'].notna()))
)

print("Invalid Q15/Q16 rows:", q16_invalid.sum())
```

```python
#to see what are those rows have to remove in Q15 → Q16 validation
clean_df.loc[q16_invalid, [
    'Q15. Recycle Waste?',
    'Q16. How Recycle?'
]]
```

```python
#q18 can contain answers like "no", "-". To normalize it.
clean_df['Q18. Monthly Income (Rs.)'] = clean_df[
    'Q18. Monthly Income (Rs.)'
].replace(
    ['-', 'No', 'Not selling'],
    np.nan
)
```

```python
#to Q17 → Q18 validation
q18_invalid = (
    ((clean_df['Q17. Income from Recyclables?'] == 'Yes') &
     (clean_df['Q18. Monthly Income (Rs.)'].isna())) |

    ((clean_df['Q17. Income from Recyclables?'] == 'No') &
     (clean_df['Q18. Monthly Income (Rs.)'].notna()))
)

print("Invalid Q17/Q18 rows:", q18_invalid.sum())
```

```python
#to see what are those rows have to remove in Q17 → Q18 validation

clean_df.loc[q18_invalid, [
    'Q17. Income from Recyclables?',
    'Q18. Monthly Income (Rs.)'
]]
```
